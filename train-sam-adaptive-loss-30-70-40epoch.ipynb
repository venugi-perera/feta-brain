```python
import os
import glob
import nibabel as nib
import numpy as np
from tqdm import tqdm

# ==========================================
# Dataset location
# ==========================================

ROOT = "/kaggle/working/feta-brain/FeTA_1.2.1_BIDS/feta1.2.1"

OUTPUT_IMAGES = "/kaggle/working/full_dataset/imagesTr"
OUTPUT_LABELS = "/kaggle/working/full_dataset/labelsTr"

os.makedirs(OUTPUT_IMAGES, exist_ok=True)
os.makedirs(OUTPUT_LABELS, exist_ok=True)

# ==========================================
# Subjects
# ==========================================

subjects = sorted(glob.glob(os.path.join(ROOT, "sub-feta*")))

print("Number of subjects:", len(subjects))

# ==========================================
# Convert
# ==========================================

for subject in tqdm(subjects):

    subject_name = os.path.basename(subject)

    # -------------------------
    # MRI
    # -------------------------
    image_file = glob.glob(
        os.path.join(subject, "anat", "*T2w.nii.gz")
    )

    if len(image_file) == 0:
        print(f"Image missing for {subject_name}")
        continue

    image_file = image_file[0]

    # -------------------------
    # Segmentation
    # -------------------------
    label_file = glob.glob(
        os.path.join(
            ROOT,
            "derivatives",
            subject_name,
            "anat",
            "*dseg.nii.gz"
        )
    )

    if len(label_file) == 0:
        print(f"Mask missing for {subject_name}")
        continue

    label_file = label_file[0]

    # -------------------------
    # Load
    # -------------------------
    image = nib.load(image_file).get_fdata().astype(np.float32)
    label = nib.load(label_file).get_fdata().astype(np.uint8)

    # Normalise MRI
    image = (image - image.min()) / (image.max() - image.min() + 1e-8)

    depth = image.shape[2]

    for i in range(depth):

        image_slice = image[:, :, i]
        label_slice = label[:, :, i]

        # Skip completely empty slices (optional)
        if np.all(label_slice == 0):
            continue

        image_name = f"{subject_name}_{i:04d}.npy"
        label_name = f"{subject_name}_{i:04d}.npy"

        np.save(
            os.path.join(OUTPUT_IMAGES, image_name),
            image_slice
        )

        np.save(
            os.path.join(OUTPUT_LABELS, label_name),
            label_slice
        )

print("Finished!")
```

```python
from torch.utils.data import Dataset
import torchvision.transforms.functional as TF
import torch
import numpy as np
import os

class FetalBrainDataset(Dataset):

    def __init__(self, image_list):
        self.image_list = image_list

    def __len__(self):
        return len(self.image_list)

    def __getitem__(self, idx):

        # --------------------------------------------------
        # File names
        # --------------------------------------------------
        image_name = self.image_list[idx]
        mask_name = image_name.replace("_0000.npy", ".npy")

        # --------------------------------------------------
        # Load image and mask
        # --------------------------------------------------
        image = np.load(
            os.path.join(IMAGE_DIR, image_name)
        ).astype(np.float32)

        mask = np.load(
            os.path.join(MASK_DIR, mask_name)
        ).astype(np.uint8)

        # --------------------------------------------------
        # Normalize image
        # --------------------------------------------------
        image = (image - image.min()) / (image.max() - image.min() + 1e-8)

        # --------------------------------------------------
        # Convert grayscale to RGB
        # --------------------------------------------------
        image = np.stack(
            [image, image, image],
            axis=0
        )

        # --------------------------------------------------
        # Convert to torch tensors
        # --------------------------------------------------
        image = torch.tensor(
            image,
            dtype=torch.float32
        )

        mask = torch.tensor(
            mask,
            dtype=torch.long
        )

        # --------------------------------------------------
        # Resize image only
        # --------------------------------------------------
        image = TF.resize(
            image,
            [1024, 1024],
            antialias=True
        )

        # --------------------------------------------------
        # Generate bounding boxes
        # --------------------------------------------------
        boxes = generate_bounding_boxes(mask.numpy())

        return image, mask, boxes
```

```python
class DiceBCELoss(nn.Module):

    def __init__(self):
        super().__init__()
        self.bce = nn.BCEWithLogitsLoss()

    def forward(self, pred, target):

        bce = self.bce(pred, target)

        pred = torch.sigmoid(pred)

        pred = pred.view(-1)
        target = target.view(-1)

        smooth = 1.0

        intersection = (pred * target).sum()

        dice = (2 * intersection + smooth) / (
            pred.sum() + target.sum() + smooth
        )

        dice_loss = 1 - dice

        return bce + dice_loss
```

```python
def generate_bounding_boxes(
    mask,
    num_classes=8,
    threshold=0,
    offset=5
):
    """
    Generate bounding boxes for each class from a mask.

    Args:
        mask: (H, W) numpy array
        num_classes: total number of classes (including background)
        threshold: mask threshold
        offset: padding around the box

    Returns:
        dict {class_id: [xmin, ymin, xmax, ymax]}
    """

    H, W = mask.shape

    bounding_boxes = {}

    for cls in range(1, num_classes):   # Skip background (0)

        binary_mask = (mask == cls)

        y_indices, x_indices = np.where(binary_mask > threshold)

        if len(x_indices) == 0 or len(y_indices) == 0:
            continue

        xmin = max(0, x_indices.min() - offset)
        xmax = min(W - 1, x_indices.max() + offset)

        ymin = max(0, y_indices.min() - offset)
        ymax = min(H - 1, y_indices.max() + offset)

        bounding_boxes[cls] = [xmin, ymin, xmax, ymax]

    # Assign empty box to missing classes
    for cls in range(1, num_classes):
        if cls not in bounding_boxes:
            bounding_boxes[cls] = [0, 0, 0, 0]

    return bounding_boxes
```

```python
def scale_box(box, from_size=256, to_size=1024):

    scale = to_size / from_size

    xmin, ymin, xmax, ymax = box

    return [
        xmin * scale,
        ymin * scale,
        xmax * scale,
        ymax * scale
    ]
```

```python
def dice_loss(pred, target, smooth=1e-6):
    pred = torch.sigmoid(pred)

    pred = pred.contiguous().view(-1)
    target = target.contiguous().view(-1)

    intersection = (pred * target).sum()

    dice = (2. * intersection + smooth) / (
        pred.sum() + target.sum() + smooth
    )

    return 1 - dice
```

```python
import torch


def adaptive_dice_bce_loss(
    pred,
    gt,
    dice_loss,
    bce_loss
):
    """
    Mild Difficulty-Aware Adaptive Dice-BCE Loss.

    Small anatomical structures receive slightly
    higher Dice weighting.

    Difficult predictions receive a small additional
    Dice weighting.

    Alpha range: 0.50 - 0.70
    Beta range : 0.50 - 0.30
    """

    ########################################################
    # FOREGROUND RATIO
    ########################################################

    foreground_ratio = gt.float().mean()

    ########################################################
    # SIZE-BASED ADAPTATION
    ########################################################

    alpha = (
        0.50
        + 0.20 * (1.0 - foreground_ratio)
    )

    ########################################################
    # PREDICTION DIFFICULTY
    ########################################################

    with torch.no_grad():

        probability = torch.sigmoid(pred)

        prediction_error = torch.abs(
            probability - gt
        ).mean()

    ########################################################
    # DIFFICULTY ADAPTATION
    ########################################################

    alpha = (
        alpha
        + 0.10 * prediction_error
    )

    ########################################################
    # LIMIT DICE WEIGHT
    ########################################################

    alpha = torch.clamp(
        alpha,
        min=0.50,
        max=0.70
    )

    ########################################################
    # BCE WEIGHT
    ########################################################

    beta = 1.0 - alpha

    ########################################################
    # LOSSES
    ########################################################

    dice = dice_loss(
        pred,
        gt
    )

    bce = bce_loss(
        pred,
        gt
    )

    ########################################################
    # FINAL LOSS
    ########################################################

    loss = (
        alpha * dice
        + beta * bce
    )

    return (
        loss,
        alpha.item(),
        beta.item()
    )
```

```python
############################################################
# TRAINING + VALIDATION
############################################################

for epoch in range(start_epoch, EPOCHS):

    print("=" * 80)
    print(f"Epoch {epoch + 1}/{EPOCHS} Started")
    print("=" * 80)

    ########################################################
    # TRAINING
    ########################################################

    print(f"Training Started - Epoch {epoch + 1}")

    sam.train()

    running_train_loss = 0.0
    running_train_alpha = 0.0
    running_train_beta = 0.0
    train_batches = 0

    pbar = tqdm(
        train_loader,
        desc=f"Training Epoch {epoch + 1}"
    )

    for image, mask, boxes in pbar:

        ####################################################
        # PREPARE DATA
        ####################################################

        image = image.squeeze(0)

        mask = mask.squeeze(0).to(device)

        ####################################################
        # RESET GRADIENTS
        ####################################################

        optimizer.zero_grad()

        ####################################################
        # IMAGE EMBEDDING
        ####################################################

        image_embedding = get_image_embedding(
            image
        )

        ####################################################
        # AVAILABLE CLASSES
        ####################################################

        available_classes = [
            cls
            for cls in range(1, NUM_CLASSES)
            if cls in boxes
        ]

        ####################################################
        # SKIP IF NO VALID CLASSES
        ####################################################

        if len(available_classes) == 0:
            continue

        ####################################################
        # RANDOMLY SELECT ONE ANATOMICAL STRUCTURE
        ####################################################

        cls = random.choice(
            available_classes
        )

        ####################################################
        # GET BOX
        ####################################################

        box = boxes[cls]

        box = prepare_box(
            scale_box(box)
        )

        ####################################################
        # PROMPT ENCODER
        ####################################################

        sparse_embeddings, dense_embeddings = (
            sam.prompt_encoder(
                points=None,
                boxes=box,
                masks=None,
            )
        )

        ####################################################
        # MASK DECODER
        ####################################################

        low_res_masks, _ = sam.mask_decoder(
            image_embeddings=image_embedding,
            image_pe=sam.prompt_encoder.get_dense_pe(),
            sparse_prompt_embeddings=sparse_embeddings,
            dense_prompt_embeddings=dense_embeddings,
            multimask_output=False,
        )

        ####################################################
        # RESIZE PREDICTION
        ####################################################

        pred = F.interpolate(
            low_res_masks,
            size=(256, 256),
            mode="bilinear",
            align_corners=False,
        )

        ####################################################
        # GROUND TRUTH
        ####################################################

        gt = (
            mask == cls
        ).float().unsqueeze(0).unsqueeze(0)

        ####################################################
        # ADAPTIVE DICE-BCE LOSS
        ####################################################

        loss, alpha, beta = adaptive_dice_bce_loss(
            pred,
            gt,
            dice_loss,
            bce_loss,
        )

        ####################################################
        # BACKPROPAGATION
        ####################################################

        scaler.scale(
            loss
        ).backward()

        scaler.step(
            optimizer
        )

        scaler.update()

        ####################################################
        # RECORD LOSS
        ####################################################

        running_train_loss += loss.item()

        ####################################################
        # RECORD ALPHA / BETA
        ####################################################

        running_train_alpha += float(alpha)

        running_train_beta += float(beta)

        train_batches += 1

        ####################################################
        # PROGRESS BAR
        ####################################################

        pbar.set_postfix(
            train_loss=f"{loss.item():.4f}",
            alpha=f"{float(alpha):.4f}",
            beta=f"{float(beta):.4f}",
            cls=cls
        )

    ########################################################
    # AVERAGE TRAINING VALUES
    ########################################################

    train_loss = (
        running_train_loss /
        max(train_batches, 1)
    )

    train_alpha = (
        running_train_alpha /
        max(train_batches, 1)
    )

    train_beta = (
        running_train_beta /
        max(train_batches, 1)
    )

    ########################################################
    # APPEND TRAINING HISTORY
    ########################################################

    train_losses.append(
        train_loss
    )

    train_alphas.append(
        train_alpha
    )

    train_betas.append(
        train_beta
    )

    ########################################################
    # TRAINING SUMMARY
    ########################################################

    print(
        f"Training Completed - Epoch {epoch + 1}"
    )

    print(
        f"Average Training Loss : "
        f"{train_loss:.6f}"
    )

    print(
        f"Average Training Alpha: "
        f"{train_alpha:.6f}"
    )

    print(
        f"Average Training Beta : "
        f"{train_beta:.6f}"
    )

    ########################################################
    # VALIDATION
    ########################################################

    print(
        f"\nValidation Started - Epoch {epoch + 1}"
    )

    sam.eval()

    running_val_loss = 0.0
    running_val_alpha = 0.0
    running_val_beta = 0.0
    val_batches = 0

    with torch.no_grad():

        pbar = tqdm(
            val_loader,
            desc=f"Validation Epoch {epoch + 1}"
        )

        for image, mask, boxes in pbar:

            ################################################
            # PREPARE DATA
            ################################################

            image = image.squeeze(0)

            mask = mask.squeeze(0).to(device)

            ################################################
            # IMAGE EMBEDDING
            ################################################

            image_embedding = get_image_embedding(
                image
            )

            ################################################
            # AVAILABLE CLASSES
            ################################################

            available_classes = [
                cls
                for cls in range(1, NUM_CLASSES)
                if cls in boxes
            ]

            ################################################
            # SKIP IF NO VALID CLASSES
            ################################################

            if len(available_classes) == 0:
                continue

            ################################################
            # RANDOM CLASS
            ################################################

            cls = random.choice(
                available_classes
            )

            ################################################
            # GET BOX
            ################################################

            box = boxes[cls]

            box = prepare_box(
                scale_box(box)
            )

            ################################################
            # PROMPT ENCODER
            ################################################

            sparse_embeddings, dense_embeddings = (
                sam.prompt_encoder(
                    points=None,
                    boxes=box,
                    masks=None,
                )
            )

            ################################################
            # MASK DECODER
            ################################################

            low_res_masks, _ = sam.mask_decoder(
                image_embeddings=image_embedding,
                image_pe=sam.prompt_encoder.get_dense_pe(),
                sparse_prompt_embeddings=sparse_embeddings,
                dense_prompt_embeddings=dense_embeddings,
                multimask_output=False,
            )

            ################################################
            # RESIZE
            ################################################

            pred = F.interpolate(
                low_res_masks,
                size=(256, 256),
                mode="bilinear",
                align_corners=False,
            )

            ################################################
            # GROUND TRUTH
            ################################################

            gt = (
                mask == cls
            ).float().unsqueeze(0).unsqueeze(0)

            ################################################
            # ADAPTIVE LOSS
            ################################################

            loss, alpha, beta = (
                adaptive_dice_bce_loss(
                    pred,
                    gt,
                    dice_loss,
                    bce_loss,
                )
            )

            ################################################
            # RECORD
            ################################################

            running_val_loss += (
                loss.item()
            )

            running_val_alpha += float(
                alpha
            )

            running_val_beta += float(
                beta
            )

            val_batches += 1

            ################################################
            # PROGRESS BAR
            ################################################

            pbar.set_postfix(
                val_loss=f"{loss.item():.4f}",
                alpha=f"{float(alpha):.4f}",
                beta=f"{float(beta):.4f}",
                cls=cls
            )

    ########################################################
    # AVERAGE VALIDATION VALUES
    ########################################################

    val_loss = (
        running_val_loss /
        max(val_batches, 1)
    )

    val_alpha = (
        running_val_alpha /
        max(val_batches, 1)
    )

    val_beta = (
        running_val_beta /
        max(val_batches, 1)
    )

    ########################################################
    # APPEND VALIDATION HISTORY
    ########################################################

    val_losses.append(
        val_loss
    )

    val_alphas.append(
        val_alpha
    )

    val_betas.append(
        val_beta
    )

    ########################################################
    # VALIDATION SUMMARY
    ########################################################

    print(
        f"Validation Completed - Epoch {epoch + 1}"
    )

    print(
        f"Average Validation Loss : "
        f"{val_loss:.6f}"
    )

    print(
        f"Average Validation Alpha: "
        f"{val_alpha:.6f}"
    )

    print(
        f"Average Validation Beta : "
        f"{val_beta:.6f}"
    )

    ########################################################
    # EPOCH SUMMARY
    ########################################################

    print("-" * 80)

    print(
        f"Epoch {epoch + 1} Summary"
    )

    print(
        f"Train Loss      : {train_loss:.6f}"
    )

    print(
        f"Train Alpha     : {train_alpha:.6f}"
    )

    print(
        f"Train Beta      : {train_beta:.6f}"
    )

    print(
        f"Validation Loss : {val_loss:.6f}"
    )

    print(
        f"Validation Alpha: {val_alpha:.6f}"
    )

    print(
        f"Validation Beta : {val_beta:.6f}"
    )

    print("-" * 80)

    ########################################################
    # CREATE OUTPUT DIRECTORY
    ########################################################

    os.makedirs(
        OUTPUT_DIR,
        exist_ok=True
    )

    ########################################################
    # SAVE LATEST CHECKPOINT
    ########################################################

    checkpoint = {

        "epoch":
            epoch + 1,

        "model_state_dict":
            sam.state_dict(),

        "optimizer_state_dict":
            optimizer.state_dict(),

        "scaler_state_dict":
            scaler.state_dict(),

        "best_val_loss":
            best_val_loss,

        ####################################################
        # LOSS HISTORY
        ####################################################

        "train_losses":
            train_losses,

        "val_losses":
            val_losses,

        ####################################################
        # ALPHA/BETA HISTORY
        ####################################################

        "train_alphas":
            train_alphas,

        "train_betas":
            train_betas,

        "val_alphas":
            val_alphas,

        "val_betas":
            val_betas,

        ####################################################
        # CURRENT VALUES
        ####################################################

        "current_train_alpha":
            train_alpha,

        "current_train_beta":
            train_beta,

        "current_val_alpha":
            val_alpha,

        "current_val_beta":
            val_beta,
    }

    torch.save(
        checkpoint,
        os.path.join(
            OUTPUT_DIR,
            "last_checkpoint.pth"
        )
    )

    print(
        "Latest checkpoint saved."
    )

    ########################################################
    # SAVE BEST MODEL
    ########################################################

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        checkpoint["best_val_loss"] = (
            best_val_loss
        )

        torch.save(
            checkpoint,
            os.path.join(
                OUTPUT_DIR,
                "best_sam_model.pth"
            )
        )

        print(
            "Best model saved."
        )

    ########################################################
    # SAVE LOSS HISTORY CSV
    ########################################################

    # All lists now have exactly the same length
    history_df = pd.DataFrame({

        "Epoch":
            range(
                1,
                len(train_losses) + 1
            ),

        "Train Loss":
            train_losses,

        "Validation Loss":
            val_losses,

        "Train Alpha":
            train_alphas,

        "Train Beta":
            train_betas,

        "Validation Alpha":
            val_alphas,

        "Validation Beta":
            val_betas,
    })

    history_df.to_csv(
        os.path.join(
            OUTPUT_DIR,
            "loss_history.csv"
        ),
        index=False
    )

    ########################################################
    # SAVE JSON
    ########################################################

    history = {

        "epochs_completed":
            epoch + 1,

        "best_val_loss":
            float(best_val_loss),

        "train_losses":
            [
                float(x)
                for x in train_losses
            ],

        "val_losses":
            [
                float(x)
                for x in val_losses
            ],

        "train_alphas":
            [
                float(x)
                for x in train_alphas
            ],

        "train_betas":
            [
                float(x)
                for x in train_betas
            ],

        "val_alphas":
            [
                float(x)
                for x in val_alphas
            ],

        "val_betas":
            [
                float(x)
                for x in val_betas
            ],
    }

    with open(
        os.path.join(
            OUTPUT_DIR,
            "loss_history.json"
        ),
        "w"
    ) as f:

        json.dump(
            history,
            f,
            indent=4
        )

    ########################################################
    # SAVE LOSS CURVE
    ########################################################

    plt.figure(
        figsize=(8, 6)
    )

    plt.plot(
        range(
            1,
            len(train_losses) + 1
        ),
        train_losses,
        marker="o",
        label="Training Loss"
    )

    plt.plot(
        range(
            1,
            len(val_losses) + 1
        ),
        val_losses,
        marker="s",
        label="Validation Loss"
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Loss"
    )

    plt.title(
        "Training and Validation Loss"
    )

    plt.grid(True)

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "loss_curve.png"
        ),
        dpi=300
    )

    plt.show()

    plt.close()

    ########################################################
    # SAVE ALPHA / BETA CURVE
    ########################################################

    plt.figure(
        figsize=(8, 6)
    )

    plt.plot(
        range(
            1,
            len(train_alphas) + 1
        ),
        train_alphas,
        marker="o",
        label="Training Alpha"
    )

    plt.plot(
        range(
            1,
            len(val_alphas) + 1
        ),
        val_alphas,
        marker="s",
        label="Validation Alpha"
    )

    plt.plot(
        range(
            1,
            len(train_betas) + 1
        ),
        train_betas,
        marker="o",
        label="Training Beta"
    )

    plt.plot(
        range(
            1,
            len(val_betas) + 1
        ),
        val_betas,
        marker="s",
        label="Validation Beta"
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Weight"
    )

    plt.title(
        "Adaptive Dice-BCE Loss Weights"
    )

    plt.grid(True)

    plt.legend()

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            OUTPUT_DIR,
            "adaptive_loss_weights.png"
        ),
        dpi=300
    )

    plt.show()

    plt.close()

    ########################################################
    # SAVE TRAINING LOG
    ########################################################

    with open(
        os.path.join(
            OUTPUT_DIR,
            "training_log.txt"
        ),
        "a"
    ) as f:

        f.write(
            f"Epoch {epoch + 1:03d} | "
            f"Train Loss: {train_loss:.6f} | "
            f"Train Alpha: {train_alpha:.6f} | "
            f"Train Beta: {train_beta:.6f} | "
            f"Val Loss: {val_loss:.6f} | "
            f"Val Alpha: {val_alpha:.6f} | "
            f"Val Beta: {val_beta:.6f} | "
            f"Best Val: {best_val_loss:.6f}\n"
        )

    print(
        "Checkpoint and logs saved successfully."
    )


############################################################
# TRAINING COMPLETED
############################################################

print("=" * 80)
print("TRAINING COMPLETED")
print("=" * 80)
```
